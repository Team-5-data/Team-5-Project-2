```python
# null and duplicate value
df.isna().sum() # we got 11 null value in postal code only in compare length of dataset droping null is easiest way out 
```

```python
df.dropna(inplace=True)
df.isna().sum() # preview
```

```python
# duplicate value
df.duplicated().sum() # no duplicate value present
```

```python
# Before date data type is object 
df["Order Date"].dtype
```

```python
df["Ship Date"].dtype
```

```python
# fixing date
# there is two date columns present we need to change there data type format into there valid format
df["Order Date"]=pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Ship Date"] =pd.to_datetime(df["Ship Date"], format="%d/%m/%Y")
```

```python
# After date data type changed to Ns datetime
df["Order Date"].dtype
```

```python
df["Ship Date"].dtype
```

```python
# dataset data types overview
df.info() # now every column has it valid datatype
```

```python
# feature engineering
# Each Customer Revenue
smp={}
for i in df["Customer ID"]:
    query_1 = df[df["Customer ID"]==i]
    smp[i]=sum(query_1["Sales"])
```

```python
cust_revenue=[]
for i in df["Customer ID"]:
    cust_revenue.append(smp[i])
    
```

```python
df["Customer_Total_Revenue"]=cust_revenue
```

```python
# Days between order date and shipping date
days =df[["Order Date" , "Ship Date"]]
df["Days_Between"]=(days["Ship Date"] - days["Order Date"]).dt.days
```
